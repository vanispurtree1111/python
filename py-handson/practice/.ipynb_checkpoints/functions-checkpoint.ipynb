```python
def player_guess():
    guess=' '
    while guess not in ['0','1','2']:
        guess= input('pick a number: 0,1 0r 2')
    return int(guess)
    
```

```python
def check_guess(mylist,guess):
    if mylist[guess]=='0':
        print("correct")
    else:
        print("wrong guess!")
        print(mylist)
```

```python
#initial list
mylist= ['','0','']
# shuffle list
mixedup_list = shuffle_of_example(mylist)
#user guess
guess = player_guess()
#check guess

check_guess(mixedup_list,guess)
```
