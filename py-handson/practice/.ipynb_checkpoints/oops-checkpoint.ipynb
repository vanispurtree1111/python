when a function is added inside the class it is called as method 
__init__ - it helps to create a instance 

```python
class Dog():
    
    def __init__(self,dogbreed,name ,spots):
        
        self.breed = dogbreed
        self.name = name
        self.spots = spots
        
```

```python
my_dog = Dog(dogbreed='Lab',name='tiny',spots=False)
```

```python
my_dog.name
```

```python
class Circle():
    pi= 3.14

    def __init__(self,radius=1):
        self.radius=radius

    def get_circumference(self):
        return self.radius * self.pi *2
```

```python
my_circle = Circle(40)
```

```python
my_circle.get_circumference()
```

```python
my_circle.radius
```

```python
class Book():
    def __init__(self,title,author,pages):
        self.title= title
        self.author=author
        self.pages=pages
    def __str__(self):
        return f"{self.title} by {self.author}"   

    def __len__(self):
        return self.pages
    def __del__(self):
        print("A book object has been deleted")
```

```python
b = Book('Python rocks','v',200)
c = Book('Learn python','vv',700)
```

```python
str(c)
```

```python
len(c)
```

```python
del b
```
