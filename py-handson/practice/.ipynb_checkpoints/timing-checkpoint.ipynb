```python
def func_one(n):
    return [str(num) for num in range(n)]
```

```python
func_one(10)
```

```python
def fun_two(n):
    return list(map(str,range(n)))
```

```python
fun_two(10)
```

```python
import time
```

```python
# current time before 
start_time = time.time()

# RUN CODE
result = func_one(1)
# current time after running the code 
end_time = time.time()

# elapsed time

elapsed_time = end_time - start_time
print(elapsed_time)
```

```python
# current time before 
start_time = time.time()

# RUN CODE
result = fun_two(1)
# current time after running the code 
end_time = time.time()

# elapsed time

elapsed_time = end_time - start_time
print(elapsed_time)
```
