```python
try:
    #block of code to attempt 
    #it may have error
    result = 10+10
except:
    #if any type of error execute this code here
    print("hey it looks like u arent adding correctly!")
else:
    print("add went well")
    print(result)
```

```python
# finally

try:
    f = open('testfile','w')
    f.write("write a test line")
except TypeError:
    print("There was a type error")
except OSError:
    print('Hey yu have a OS Error')
except:
    print('All other exceptionS!')
finally:
    print("i always run no matter what")
```

```python
def ask_for_int():
    try:
        result =int(input("please provide numner"))
    except:
        print("Whoops! That is not a number")
    finally:
        print("End os try/except/finally")
```
