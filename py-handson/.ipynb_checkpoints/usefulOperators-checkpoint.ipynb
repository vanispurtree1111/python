enumerate - return the index along with the object
useful in showing the letters along with its index

```python
word = 'hello world'

for letter in enumerate(word):
    print(letter)
```

```python
word = 'hello world'

for index,letter in enumerate(word):
    print(letter)
    print(index)
    print('\n')
```

zip ------------------------------ pair and give the items in tuple

```python
mylist1 = [1,2,3]
mylist2 =['a','b','c']
```

```python
for item in zip(mylist1,mylist2):
    print(item)
# pair and give the items in tuple
```
