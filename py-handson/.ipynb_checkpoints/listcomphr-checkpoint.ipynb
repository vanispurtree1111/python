list comprehensions are a unique way of quickly creating a list wuth python.
if we are using a for loop along with .append() to create a list 

```python
mystring ='hello'
```

```python
mylist= []
for letter in mystring:
    mylist.append(letter)
```

```python
mylist
```

 by using for loop as well can able to achieve this 

```python
myList = [letter for letter in mystring]
```

```python
myList
```

```python
mylist = [x for x in 'word']
# element for element in sting - here is the syntax for it 
```

```python
mylist
```

how to use if and else in list comprehesion

```python
results = [x if x%2==0 else'ODD' for x in range(0,11)]
```

```python
results
```

nested loops 

```python
mylist = []
for x in [2,4,6]:
    for y in [10,30,20]:
        mylist.append(x*y)
        
```

```python
mylist
```
