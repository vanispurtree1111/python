Dictionaries along with its functions 
can able to fetch the data of items without the use of indexing as we have key and values to add 

```python
my_Dict = { "name" : "vani", "age":25,"place":"Tumkur"}
```

```python
my_Dict
```

```python
 my_Dict['name']
# by passing the key we can get the value
```

```python
price_lookup = { 'apple':400, 'orange':200,'banana':100,'pineapple':["it is fruit",200,"good for health"]}
```

```python
price_lookup
```

```python
price_lookup['pineapple']
```

```python
price_lookup['pineapple'][2]
```

```python
price_lookup['pineapple'][2].upper()
```

```python
# instead of adding so many steps like above, we can modify in one line condition and get the same output as we needed
price_lookups = { 'apple':400, 'orange':200,'banana':100,'pineapple':["it is fruit",200,"good for health"]}
```

```python
price_lookups['pineapple'][0].upper()
```

```python
# To append/ add new key value in dict
exam = {'k1':100,'k2':200,'k3':300}
```

```python
exam['k4'] = 400
```

```python
exam
```

```python
# to change the value of the already existing one value 
exam['k1']=500
exam
```

to see all the keys here is the command 
syntax: dictionaryname.keys()

```python
exam.keys()
```

to see all the key along with the values then here is the command 
syntax: dictionaryname.items()

```python
exam.items()
```
