while loops :

    syntax :
while some_boolen_condition:
    #do something
else:
    #do something
    

```python
x = 0
while x<5:
    print(f'the current value of x is {x}')
    x = x+1
else:
    print("X is not less than 5")
```

break ,continue ,pass
break: Breaks out of the current enclosing loop:
continue : goes to the top of the closest enclosing loop.
pass: Does nothing at all

```python
x = [1,2,3]
for item in x:
    pass    
    # used it kind of placeholder to not get any errrors
print('end of my script')
```

```python
mylist = 'helloWorld'
for letter in mylist:
    if letter == 'h':
        continue
    print(letter)
```

```python
mylist = 'helloWorld'
for letter in mylist:
    if letter == 'W':   
       break
    print(letter)
```
